Напомним суть метода A2C:

Есть 2 нейронный сети: Actor и Critic.

Critic - по-сути функция, оценивающая ожидаемую награду, которую обучают под дискретное уравнение Беллмана.

Actor - политика, которую мы обучаем максимизировать ожидаемую награду.



В классическом A2C лоссы для них выглядят следующим образом:

$L_{actor} = -\frac{1}{MN} \sum_{m=0}^M \sum_{t=0}^N log(π(a_t^m|s_t^m))A(a_t^m, s_t^m)$

$L_{critic} = \frac{1}{MN} \sum_{m=0}^M \sum_{t=0}^N (V(s_t^m) - y(s_t^m, a_t^m))$

$y(s_t^m, a_t^m) = \sum_{t'=t}^{t''} γ^{t'-t} r(s_{t'}, a_{t'}) + γ^{t''-t+1} V(s_{t''})$, где $t\leq t''<N-1$
